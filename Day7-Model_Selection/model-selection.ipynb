```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('../datasets/final_dataset_missing_value_imputed_v5.csv')
```

```python
df.head()
```

```python
# undoing the conversion of numeric columns to categorical
df['furnishing_type'].value_counts()
```

```python
df['furnishing_type'] = df['furnishing_type'].replace({0.0: 'unfurnished', 1.0: 'semi-furnished', 2.0: 'furnished'})
```

```python
df.head()
```

```python
def categorize_facility_score(score):
    if(0 <= score < 50):
        return "Low"
    elif(50 <= score < 150):
        return "Medium"
    elif 150 <= score < 175:
        return "High"
    else:
        return None
    
df['facility_category'] = df['facility_score'].apply(categorize_facility_score)
```

```python
df.head()
```

```python
def floor_range(floor):
    if 0 <= floor <= 2:
        return "Low Floor"
    elif 3 <= floor <= 10:
        return "Mid Floor"
    elif 11 <= floor <= 51:
        return "High Floor"
    else:
        return None

df['floor_category'] = df['floorNum'].apply(floor_range)
```

```python
df.head()
```

```python
X = df.drop(columns=['society', 'price_per_sqft', 'facility_score', 'floorNum', 'price', 'property_id', 'others', 'study room'])
y = df['price']
```

```python
X.head()
```

```python
y_transformed = np.log1p(y) # log transformation to reduce skewness
```

### Ordinal Encoding

```python
cols_to_encode = ['property_type', 'sector', 'balcony', 'agePossession', 'furnishing_type', 'facility_category', 'floor_category']
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder
preprocessor = ColumnTransformer(
    transformers = [
        ('num', StandardScaler(), ['bedRoom', 'bathroom', 'builtup_area', 'servant room', 'store room']),
        ('cat', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), cols_to_encode)
    ],
    remainder='passthrough'
)
```

```python
# pipeline creation
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])
```

```python
from sklearn.model_selection import cross_val_score, KFold
kf = KFold(n_splits=15, shuffle=True, random_state=42)
scores = cross_val_score(pipeline, X, y_transformed, cv=kf, scoring='r2')
```

```python
scores.mean(), scores.std()
```

```python
scores
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y_transformed, test_size=0.2, random_state=42)
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
y_pred = np.expm1(y_pred)  # inverse log transformation
```

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(np.expm1(y_test), y_pred)
mae
```

```python
# function to test each algorithm
def testing(model_name, model):
    output = []
    output.append(model_name)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('regressor', model)
    ])
    
    # kfolds
    kf = KFold(n_splits=15, shuffle=True, random_state=42)
    scores = cross_val_score(pipeline, X, y_transformed, cv=kf, scoring='r2')
    
    output.append(scores.mean())
    
    # fit the model
    X_train, X_test, y_train, y_test = train_test_split(X, y_transformed, test_size=0.2, random_state=42)
    pipeline.fit(X_train, y_train)
    
    y_pred = pipeline.predict(X_test)
    y_pred = np.expm1(y_pred)  # inverse log transformation
    
    output.append(mean_absolute_error(np.expm1(y_test), y_pred))
    return output
```

```python
from sklearn.linear_model import Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, AdaBoostRegressor, ExtraTreesRegressor
from xgboost import XGBRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.svm import SVR

models_dict = {
    'linear_regression': LinearRegression(),
    'svr': SVR(),
    'ridge': Ridge(),
    'lasso': Lasso(),
    'decision_tree': DecisionTreeRegressor(),
    'random_forest': RandomForestRegressor(),
    'gradient_boosting': GradientBoostingRegressor(),
    'adaboost': AdaBoostRegressor(),
    'xgboost': XGBRegressor(),
    'mlp': MLPRegressor(), # multi layer perceptron neural network
    'extra_trees': ExtraTreesRegressor()
}
```

```python
model_output = []
for model_name, model in models_dict.items():
    output = testing(model_name, model)
    model_output.append(output)
    
model_output
```

```python
pd.DataFrame(model_output, columns=['name', 'r2', 'mae']).sort_values(['mae'])
```

### OHE with PCA

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), ['bedRoom', 'bathroom', 'builtup_area', 'servant room', 'store room']),
        ('cat', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), cols_to_encode), # categorical cols where order matters
        ('cat01', OneHotEncoder(handle_unknown='ignore'), ['sector', 'agePossession', 'furnishing_type'])
    ],
    remainder='passthrough'
)
```

```python
from sklearn.decomposition import PCA

pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('pca', PCA(n_components=.95)),
    ('regressor', LinearRegression())
])
```

```python
kfold = KFold(n_splits=15, shuffle=True, random_state=42)
scores = cross_val_score(ohe_pipeline, X, y_transformed, cv=kfold, scoring='r2')

scores.mean()
```

```python
# function to test each algorithm
def scorer(model_name, model):
    
    output = []
    
    output.append(model_name)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('pca', PCA(n_components=0.95)),
        ('regressor', model)
    ])
    
    # K-fold cross-validation
    kfold = KFold(n_splits=10, shuffle=True, random_state=42)
    scores = cross_val_score(pipeline, X, y_transformed, cv=kfold, scoring='r2')
    
    output.append(scores.mean())
    
    X_train, X_test, y_train, y_test = train_test_split(X,y_transformed,test_size=0.2,random_state=42)
    
    pipeline.fit(X_train,y_train)
    
    y_pred = pipeline.predict(X_test)
    
    y_pred = np.expm1(y_pred)
    
    output.append(mean_absolute_error(np.expm1(y_test),y_pred))
    
    return output
```

```python
model_dict = {
    'linear_reg':LinearRegression(),
    'svr':SVR(),
    'ridge':Ridge(),
    'LASSO':Lasso(),
    'decision tree': DecisionTreeRegressor(),
    'random forest':RandomForestRegressor(),
    'extra trees': ExtraTreesRegressor(),
    'gradient boosting': GradientBoostingRegressor(),
    'adaboost': AdaBoostRegressor(),
    'mlp': MLPRegressor(),
    'xgboost':XGBRegressor()
}
```

```python
model_output = []
for model_name,model in model_dict.items():
    model_output.append(scorer(model_name, model))
```

### target encoding

when high dimensional data is present, we use target encoding.  
it encodes the categorical values based on the target (price) variable.

get to the documentation for better understanding.

```python
import category_encoders as ce

cols_to_encode = ['property_type', 'sector', 'balcony', 'agePossession', 'furnishing_type', 'facility_category', 'floor_category']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), ['bedRoom', 'bathroom', 'builtup_area', 'servant room', 'store room']),
        ('cat', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), cols_to_encode),
        ('cat1', OneHotEncoder(handle_unknown='ignore', drop='first', sparse_output=False), ['agePossession']),   
        ('target_enc', ce.TargetEncoder(), ['sector'])
    ],
    remainder='passthrough'
)
```

```python
pipeline = Pipeline(
    [
        ('preprocessor', preprocessor),
        ('regressor', LinearRegression())
    ]
)
```

```python
kfold = KFold(n_splits=15, shuffle=True, random_state=42)
scores = cross_val_score(pipeline, X, y_transformed, cv=kfold, scoring='r2')

scores.mean(), scores.std()
```

```python
def scorer(model_name, model):
    output = []
    
    output.append(model_name)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('regressor', model)
    ])
    
    # K-fold cross-validation
    kfold = KFold(n_splits=10, shuffle=True, random_state=42)
    scores = cross_val_score(pipeline, X, y_transformed, cv=kfold, scoring='r2')
    output.append(scores.mean())
    
    X_train, X_test, y_train, y_test = train_test_split(X, y_transformed, test_size=0.2, random_state=42)
    pipeline.fit(X_train, y_train)
    
    y_pred = pipeline.predict(X_test)
    y_pred = np.expm1(y_pred)  # inverse log transformation
    
    
    output.append(mean_absolute_error(np.expm1(y_test), y_pred))
    return output
```

```python
model_output = []
for model_name, model in model_dict.items():
    output = scorer(model_name, model)
    model_output.append(output)

pd.DataFrame(model_output, columns=['name', 'r2', 'mae']).sort_values(['mae']).sort_values(['mae'])
```

target encoding works better on tree based models.  
slight improvements but not on linear models.
